```python
from ultralytics import YOLO
import torch

if torch.cuda.is_available():
    device = 0  # GPU (CUDA:0)
    print("✅ CUDA GPU is available. Using GPU.")
else:
    device = 'cpu'
    print("⚠️ CUDA GPU not available. Using CPU.")
```

```python
from ultralytics import YOLO
import torch
!wget https://github.com/ultralytics/assets/releases/download/v0.0.0/yolov8n.pt -O yolov11n.pt
model=YOLO('yolov11n.pt')

# model = YOLO('/content/drive/MyDrive/my_model.pt')
```

```python

results = model.train(
    data='/content/drive/MyDrive/Vehicle/data.yaml',
    epochs=100,
    imgsz=640,
    batch=16,
    name='model_v11',
    device=device,
    verbose=True,
    val=True,
    plots=True,
    save=True,
    save_period=10,
    workers=6,
    exist_ok=True,
    cache=True,
    patience=10
)
```

```python
model.save('My_model')
```

```python

metrics = model.val(
    data='/content/drive/MyDrive/Vehicle/data.yaml',
    split='test',
    imgsz=640,
    batch=16,
    device=device,
    save_json=True,
    verbose=True,
    plots=True
)
```
